```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
n = 1000000
defin = stats.uniform.rvs(3,14,n)
extr = stats.expon.rvs(scale = 10, size = n)
eda = stats.norm.rvs(25,5,n)
dev = stats.norm.rvs(5,1,n)
test = stats.norm.rvs(65,2,n)
design = stats.expon.rvs(loc=20, scale = 20, size = n)
pipe = stats.norm.rvs(30,5,n)
logging = stats.uniform.rvs(10,20,n)
test = stats.norm.rvs(30,6,n)
```

```python
devel = eda + dev + test
prod = design + pipe + logging
print((devel > prod).sum()/n)
# percent where development takes longer than production
```

```python
days = defin + extr + np.amax(np.column_stack((devel,prod)),1)+test
days
```

```python
print(np.mean(days), np.std(days), np.quantile(days,0.95))
```

```python
# (d) with failures
tests = stats.geom.rvs(0.5, size = n)
test2 = np.zeros(n)
for i in range(n):
    test2[i] = np.sum(stats.norm.rvs(loc=30, scale=6, size=tests[i]))
```

```python
days2 = defin + extr + np.amax(np.column_stack((devel,prod)),1)+test2
print(np.mean(days2), np.std(days2), np.quantile(days2,0.95))
```
